**Project Requirements**  
1. Data Import & Processing

Your application should:  

Read data from the CSV file.  
Display the dataset.  
Display dataset information.  
Handle missing (null) values appropriately.  
Detect and handle zero values where necessary.  
Generate descriptive statistics.  
Create additional calculated columns such as:  
* Average
* Total Marks  
* Grade  
* Rank  
2. Dashboard Overview

The dashboard should provide an overview including:

Total Students  
Average Attendance  
Highest Average Marks  
Lowest Average Marks  
Class Average  
Number of Students Above 85%  
Number of Students Below 60%  

Display these as summary cards or KPI widgets.

3. Student Analysis  

The application should allow users to:  
* View complete information for each student.
* Compare students.
* Identify the class topper.
* Identify the lowest-performing student.
* Display student rankings.
* Filter students based on marks or attendance.
4. Data Visualizations

Your dashboard should include multiple visualizations using different Python libraries.

Possible visualizations include:

* Bar Charts  
* Line Charts  
* Pie Charts  
* Scatter Plots  
* Histograms  
* Box Plots  
* Heatmaps  
* Area Charts  
* Bubble Charts  
* Radar Charts (Optional)  
* Sunburst Charts (Optional)  
* Treemaps (Optional)  

Use different libraries wherever possible instead of relying on a single library.

5. Interactive Features

Your dashboard should include interactive components such as:

* Student selection dropdown  
* Subject selection  
* Filters  
* Search functionality  
* Hover information  
* Zoom and pan  
* Dynamic chart updates  
* Sort options  

6. Performance Insights

Generate meaningful insights such as:

* Top 5 performers  
* Bottom 5 performers  
* Attendance vs Performance  
* Subject-wise average  
* Highest scoring subject  
* Lowest scoring subject  
* Correlation between attendance and marks  
* Students with perfect attendance  
* Students requiring improvement  

7. Data Cleaning Module

Your project should clearly demonstrate:

* Missing value detection
* Missing value handling
* Duplicate detection
* Duplicate removal (if any)
* Outlier detection
* Data validation
8. Export Functionality

The dashboard should allow users to:

* Export the processed dataset as CSV.
* Save charts as images (optional).
* Save the dashboard as an HTML file (if applicable).
9. User Interface

The dashboard should have:

* Modern layout
* Professional colour scheme
* Clear navigation
* Responsive design
* Proper chart titles
* Axis labels
* Legends
* Tooltips
* Icons (optional)
* Consistent spacing and alignment

Focus on creating an interface that is easy to understand and visually appealing.

Additional Features

You are encouraged to implement additional features beyond the requirements, such as:

* Dark Mode / Light Mode  
* Login Screen  
* Multiple Dashboard Pages  
* AI-generated insights  
* Download Report (PDF)  
* Animated Charts  
* Live Search  
* Theme Switcher
